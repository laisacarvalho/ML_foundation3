```python
# Importanto as bibliotecas
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn import metrics
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.metrics import mean_squared_error
from math import sqrt
```

```python
# Importando o dataset
data = datasets.load_breast_cancer() #o Python carrega uma base de dados inteira dentro da variável data.
data.feature_names
```

```python
print(data.DESCR)
```

```python
data.target #É a resposta correta para cada linha (Y real)

data.data #É a parte que contém as características dos tumores (X).
```

```python
#Exercicio 1 Separe o dataset em conjunto de treino e teste, sendo 70% para treino e 30% para teste
## Divide o dataset em treino (70%) e teste (30%).
# X = características dos tumores (entradas)
# y = diagnóstico (saída: benigno ou maligno)
# X_train e y_train serão usados para treinar o modelo.
# X_test e y_test serão usados para avaliar se o modelo aprendeu corretamente.

X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.3, random_state=42)
```

```python
#Exercicio 2 Treine um classificador SVM usando kernel linear e outro usando kernel RBF
# Antes de treinar precisa normalizar os dados para que todas as variáveis fiquem na mesma escala.
#fit_transform() calcula a média e o desvio padrão do conjunto de treino
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Treinar Kernel linear
svm_linear = svm.SVC(kernel='linear')
svm_linear.fit(X_train, y_train)

#Treinar Kernel rbf
svm_rbf = svm.SVC(kernel='rbf')
svm_rbf.fit(X_train, y_train)
```

```python
# Exercicio 3 Faça a predição para o conjunto de teste usando os dois kernels
y_pred_linear = svm_linear.predict(X_test)
y_pred_rbf = svm_rbf.predict(X_test)
print("Valores reais:", y_test[:10])
print("Linear:", y_pred_linear[:10])
print("RBF:", y_pred_rbf[:10])
```

```python
# Exercicio 4 Obtenha a Acurácia, Precision e Recall para cada kernel
accuracy_linear = metrics.accuracy_score(y_test, y_pred_linear)
precision_linear = metrics.precision_score(y_test, y_pred_linear)
recall_linear = metrics.recall_score(y_test, y_pred_linear)
print(accuracy_linear)
print(precision_linear)
print(recall_linear)
```
